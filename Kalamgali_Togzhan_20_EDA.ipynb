## 03. Деректердің сапасы

**Мақсаты:** бос мәндерді, қайталанған жолдарды, ақпарат бермейтін бағандарды және қате типтерді тексеріп, деректерді тазалау.

### Бос мәндер және қайталанулар

```python
# Әр бағандағы бос мәндер саны және пайызы бір кестеде
missing = pd.DataFrame({
    "бос_саны": df.isna().sum(),
    "бос_пайызы": (df.isna().mean() * 100).round(2),
})

print("Барлық бос мәндер саны:", missing["бос_саны"].sum())
missing.sort_values("бос_саны", ascending=False).head(10)
```

```python
# Толық қайталанған жолдар саны
print("Қайталанған жолдар:", df.duplicated().sum())

# Қызметкер нөмірі бірегей ме
print("EmployeeNumber бірегей ме:", df["EmployeeNumber"].is_unique)
```

### Ақпарат бермейтін бағандар

```python
# Тек бір ғана мәні бар бағандар талдауға ешқандай ақпарат бермейді
const_cols = [c for c in df.columns if df[c].nunique() == 1]

for c in const_cols:
    print(c, "→", df[c].unique().tolist())
```

### Тазалау және типтерді түрлендіру

```python
# Бастапқы кестені өзгертпеу үшін көшірме жасаймыз
df_clean = df.copy()

# 1) Тұрақты бағандарды алып тастаймыз
df_clean = df_clean.drop(columns=const_cols)

# 2) Attrition бағанынан 0/1 белгісін жасаймыз: үлесті mean() арқылы есептеу үшін
df_clean["AttritionFlag"] = (df_clean["Attrition"] == "Yes").astype(int)

# 3) Мәтіндік бағандарды category типіне түрлендіреміз
for c in ["Attrition", "BusinessTravel", "Department", "EducationField",
          "Gender", "JobRole", "MaritalStatus", "OverTime"]:
    df_clean[c] = df_clean[c].astype("category")

# 4) Education кодтарын түсінікті атауларға аударамыз (реттік санат)
edu_map = {1: "Below College", 2: "College", 3: "Bachelor", 4: "Master", 5: "Doctor"}
df_clean["EducationLevel"] = pd.Categorical(df_clean["Education"].map(edu_map),
                                            categories=list(edu_map.values()), ordered=True)

print("Жаңа пішіні:", df_clean.shape)
df_clean.dtypes.astype(str).value_counts()
```

```python
# Логикалық тексеру: компаниядағы өтіл жалпы еңбек өтілінен аспауы керек
bad = df_clean[df_clean["YearsAtCompany"] > df_clean["TotalWorkingYears"]]
print("Логикалық қайшылығы бар жолдар:", len(bad))
```

**Шешімнің негіздемесі.** Деректерде бірде-бір бос мән және бірде-бір қайталанған жол жоқ, сондықтан жою немесе толтыру қажет болмады. `EmployeeCount` (барлығы 1), `Over18` (барлығы Y) және `StandardHours` (барлығы 80) бағандарында бір ғана мән бар, олар ешқандай айырмашылық көрсетпейді, сондықтан алынып тасталды. Типі қате оқылған сан немесе күн бағаны жоқ, бірақ мәтіндік бағандар `category` типіне ауыстырылды, `Attrition` үшін 0/1 белгісі (`AttritionFlag`) қосылды, ал `Education` кодтары атауларға аударылды. `YearsAtCompany > TotalWorkingYears` сияқты логикалық қайшылық табылмады.

## 04. Сипаттамалық статистика

**Мақсаты:** `describe()` арқылы сандық бағандардың статистикасын көру, негізгі 3 бағанның орташасы мен медианасын салыстыру және шектен тыс мәндерді анықтау.

```python
# Барлық сандық бағандардың статистикасы (транспонирленген түрде оқуға ыңғайлы)
df_clean.describe().T.round(2)
```

### Орташа мен медиананы салыстыру

```python
# EDA бағыты үшін ең маңызды 3 сандық баған
key_num = ["MonthlyIncome", "Age", "YearsAtCompany"]

cmp = pd.DataFrame({
    "орташа": df_clean[key_num].mean(),
    "медиана": df_clean[key_num].median(),
    "min": df_clean[key_num].min(),
    "max": df_clean[key_num].max(),
}).round(1)

# Орташа медианадан неше пайызға үлкен
cmp["айырма_%"] = ((cmp["орташа"] - cmp["медиана"]) / cmp["медиана"] * 100).round(1)
cmp
```

### Шектен тыс мәндер

```python
# IQR әдісімен шектен тыс мәндерді анықтау
for c in key_num:
    q1, q3 = df_clean[c].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr   # төменгі шек
    upper = q3 + 1.5 * iqr   # жоғарғы шек
    n_out = ((df_clean[c] < lower) | (df_clean[c] > upper)).sum()
    print(f"{c:15s} шектер: [{lower:.0f}; {upper:.0f}]  шектен тыс: {n_out}")
```

```python
# Компанияда 40 жыл істеген адам шынайы ма: жасы мен жалпы өтілін тексереміз
df_clean.nlargest(3, "YearsAtCompany")[["Age", "TotalWorkingYears", "YearsAtCompany", "JobRole"]]
```

**Түсіндірме.** `MonthlyIncome` бойынша орташа (≈6503) медианадан (4919) шамамен 32%-ға жоғары: табыс оңға қисайған, яғни көпшілігі аз табыс алады, ал аз ғана басшылардың жалақысы орташаны көтереді. `YearsAtCompany` бойынша да орташа (7.0) медианадан (5) жоғары, ал `Age` симметриялы (36.9 және 36). Ең кіші және ең үлкен мәндер шынайы: жас 18–60, табыс 1009–19999, компанияда 40 жыл істеген адам 58 жаста және оның жалпы өтілі де 40 жыл. IQR бойынша табыста 114 және компаниядағы жылдарда 104 «шектен тыс» мән бар, бірақ олар қате емес — жоғары лауазымдағы және ұзақ істеген қызметкерлер, сондықтан оларды жоймаймыз.

## 06. Жаңа бағандар

**Мақсаты:** табыс пен кетуді талдауға пайдалы жаңа белгілер құру: біреуі есептелген сан, екіншісі `pd.cut()` арқылы санаттар.

```python
# 1) Есептелген сан: еңбек өтілінің бір жылына шаққандағы айлық табыс
#    (+1 нөлге бөлуді болдырмау үшін, өйткені өтілі 0 жыл адамдар бар)
df_clean["IncomePerExpYear"] = (df_clean["MonthlyIncome"] /
                                (df_clean["TotalWorkingYears"] + 1)).round(1)

# 2) Сандық бағанды санаттарға бөлу: айлық табыс деңгейі
df_clean["IncomeBand"] = pd.cut(df_clean["MonthlyIncome"],
                                bins=[0, 3000, 6000, 10000, 20000],
                                labels=["Low (<3k)", "Medium (3–6k)", "High (6–10k)", "Very high (>10k)"])

# 3) Қосымша санат: жас тобы
df_clean["AgeGroup"] = pd.cut(df_clean["Age"], bins=[17, 25, 35, 45, 60],
                              labels=["18–25", "26–35", "36–45", "46–60"])

df_clean[["MonthlyIncome", "TotalWorkingYears", "IncomePerExpYear", "IncomeBand", "Age", "AgeGroup"]].head()
```

```python
# Жаңа санаттардың таралуы
print(df_clean["IncomeBand"].value_counts().sort_index())
print()
print(df_clean["AgeGroup"].value_counts().sort_index())
```

**Не үшін қажет:**

- `IncomePerExpYear` — тәжірибесіне қатысты «аз төленетін» қызметкерлерді анықтауға көмектеседі: бірдей табыста тәжірибесі көп адам өзін бағаланбаған деп сезінуі мүмкін.
- `IncomeBand` — табысты 4 деңгейге бөліп, әр деңгейдегі кету үлесін салыстыруға мүмкіндік береді.
- `AgeGroup` — жас топтары бойынша кетуді салыстыру үшін қосымша баған.

Табыс санаттарында ең көп топ Medium (519 адам), ал Low тобында 395 адам бар.

## 07. Фильтрлеу

**Мақсаты:** EDA бағыты бойынша нақты сұрақтар қойып, оларға бір шартты, бірнеше шартты (`&`), `isin()`, `between()` және `query()` сүзгілері арқылы жауап беру.

### Сұрақ 1 (бір шарт): үстеме уақыт жұмыс істейтін қызметкерлердің кету үлесі қандай?

```python
# Бір шарт: OverTime == "Yes"
ot = df_clean[df_clean["OverTime"] == "Yes"]

print("Үстеме жұмыс істейтіндер:", len(ot))
print(f"Олардың кету үлесі: {ot['AttritionFlag'].mean():.1%}")
print(f"Жалпы кету үлесі: {df_clean['AttritionFlag'].mean():.1%}")
```

**Жауап:** үстеме уақыт жұмыс істейтін 416 қызметкердің 30.5%-ы кеткен — бұл жалпы деңгейден (16.1%) шамамен 2 есе жоғары.

### Сұрақ 2 (бірнеше шарт, `&`): айлық табысы 3000-нан төмен Sales Representative-тердің қаншасы кетті?

```python
# Екі шарт: лауазым ЖӘНЕ төмен табыс
low_sr = df_clean[(df_clean["JobRole"] == "Sales Representative") &
                  (df_clean["MonthlyIncome"] < 3000)]

print("Мұндай қызметкерлер:", len(low_sr))
print("Кеткендер:", low_sr["AttritionFlag"].sum())
print(f"Кету үлесі: {low_sr['AttritionFlag'].mean():.1%}")
```

**Жауап:** 68 төмен табысты Sales Representative-тің 28-і, яғни 41.2%-ы кеткен — бұл деректердегі ең «қауіпті» топтардың бірі.

### Сұрақ 3 (`isin()`): жоғары басшылық лауазымдарда (Manager, Research Director) кету үлесі қандай?

```python
# isin(): мән берілген тізімде бар ма
top_mgmt = df_clean[df_clean["JobRole"].isin(["Manager", "Research Director"])]

print("Басшылар саны:", len(top_mgmt))
print("Кеткендер:", top_mgmt["AttritionFlag"].sum())
print(f"Кету үлесі: {top_mgmt['AttritionFlag'].mean():.1%}")
```

**Жауап:** 182 басшының тек 7-еуі (3.8%) кеткен — жалпы деңгейден 4 есе төмен.

### Сұрақ 4 (`between()`): 18–25 жастағы жас қызметкерлердің кету үлесі қандай?

```python
# between(): шекаралар нәтижеге кіреді
young = df_clean[df_clean["Age"].between(18, 25)]

print("18–25 жастағылар:", len(young))
print(f"Кету үлесі: {young['AttritionFlag'].mean():.1%}")
```

**Жауап:** 123 жас қызметкердің 35.8%-ы кеткен — жас мамандар компанияда ең аз тұрақтайды.

### Сұрақ 5 (`query()`): 1-деңгейдегі, үстеме жұмыс істейтін және жұмысқа қанағаттануы төмен қызметкерлердің кету үлесі қандай?

```python
# query(): шартты мәтін түрінде жазу
risk = df_clean.query("JobLevel == 1 and OverTime == 'Yes' and JobSatisfaction == 1")

print("Мұндай қызметкерлер:", len(risk))
print(f"Кету үлесі: {risk['AttritionFlag'].mean():.1%}")
```

**Жауап:** мұндай 34 қызметкер бар және олардың 55.9%-ы кеткен — кіші лауазым, үстеме жұмыс және төмен қанағаттану бірге болғанда кету қаупі жалпы деңгейден 3,5 есе жоғары.

## 08. Топтау

**Мақсаты:** `groupby()` және `agg()` арқылы лауазымдарды кету үлесі, табыс және өтіл бойынша бір кестеде салыстыру.

### Лауазымдар бойынша жиынтық көрсеткіштер

```python
# JobRole бойынша топтап, бірнеше көрсеткішті атаулы агрегация арқылы есептейміз
role_stats = df_clean.groupby("JobRole", observed=True).agg(
    қызметкер_саны=("AttritionFlag", "count"),
    кету_үлесі=("AttritionFlag", "mean"),
    табыс_орташа=("MonthlyIncome", "mean"),
    табыс_медиана=("MonthlyIncome", "median"),
    табыс_min=("MonthlyIncome", "min"),
    табыс_max=("MonthlyIncome", "max"),
    өтіл_орташа=("YearsAtCompany", "mean"),
).sort_values("кету_үлесі", ascending=False)   # кету үлесі бойынша сұрыптау

# Үлесті пайызға айналдырамыз
role_stats["кету_үлесі"] = (role_stats["кету_үлесі"] * 100).round(1)
role_stats = role_stats.round(1)
role_stats
```

### Кеткендер мен қалғандарды салыстыру

```python
# Кеткен және қалған қызметкерлердің орташа және медиана көрсеткіштері
df_clean.groupby("Attrition", observed=True)[["MonthlyIncome", "Age", "YearsAtCompany", "JobLevel"]] \
        .agg(["mean", "median"]).round(1)
```

### Табыс санаттары бойынша кету үлесі

```python
income_stats = df_clean.groupby("IncomeBand", observed=True).agg(
    саны=("AttritionFlag", "count"),
    кеткендер=("AttritionFlag", "sum"),
    кету_үлесі=("AttritionFlag", "mean"),
)
income_stats["кету_үлесі"] = (income_stats["кету_үлесі"] * 100).round(1)
income_stats
```

**Түсіндірме.** Sales Representative ерекше көзге түседі: кету үлесі 39.8%, орташа табысы ең төмен (≈2626) және компаниядағы орташа өтілі тек 2.9 жыл. Керісінше, Research Director (2.5%) мен Manager (4.9%) — ең жоғары табысты (≈16 000–17 000) және ең тұрақты лауазымдар. Кеткендердің медианалық табысы 3202, ал қалғандарда 5204; кеткендер жасырақ (медиана 32 және 36 жас). Табыс санаты өскен сайын кету үлесі тұрақты төмендейді: Low (<3k) — 28.6%, Very high (>10k) — 8.9%.

## 09. Екі белгі бойынша жиынтық кесте

**Мақсаты:** `pd.pivot_table()` арқылы лауазым мен үстеме жұмыстың кету үлесіне бірлескен әсерін көрсету.

```python
# Лауазым × Үстеме жұмыс бойынша кету үлесі (%)
pivot_ot = pd.pivot_table(df_clean, values="AttritionFlag",
                          index="JobRole", columns="OverTime",
                          aggfunc="mean", observed=True) * 100
pivot_ot = pivot_ot.round(1)

# Үстеме жұмыс кету үлесіне қанша пайыз «қосады»
pivot_ot["айырма"] = (pivot_ot["Yes"] - pivot_ot["No"]).round(1)
pivot_ot = pivot_ot.sort_values("Yes", ascending=False)
pivot_ot
```

```python
# Кестедегі ең жоғары және ең төмен мәндер
vals = pivot_ot[["No", "Yes"]].stack()

print("Ең жоғары:", vals.idxmax(), "→", vals.max(), "%")
print("Ең төмен :", vals.idxmin(), "→", vals.min(), "%")
```

```python
# Қосымша: лауазым × табыс санаты бойынша кету үлесі (NaN = мұндай топ жоқ)
pivot_inc = pd.pivot_table(df_clean, values="AttritionFlag", index="JobRole",
                           columns="IncomeBand", aggfunc="mean", observed=True) * 100
pivot_inc.round(1)
```

**Түсіндірме.** Ең жоғары мән — үстеме жұмыс істейтін Sales Representative: олардың 66.7%-ы кеткен; ең төмен мән — үстеме жұмыс істемейтін Manager: тек 1.3%. Healthcare Representative-тен басқа барлық лауазымда үстеме жұмыс кету үлесін арттырады, әсіресе Laboratory Technician (15.7% → 50.0%) және Research Scientist (7.2% → 34.0%) топтарында. Яғни табысы төмен лауазым мен шамадан тыс жүктеменің қосындысы — кетудің ең күшті белгісі.
